```python
import pandas as pd
import os

def carregar_arquivos(caminho_ou_url=None, **kwargs):
    """
    Carrega automaticamente arquivos de diferentes formatos e exibe as 10 primeiras linhas.
    Se nenhum caminho/URL for informado, solicita o input do usuário na tela.

    Formatos suportados: CSV, TSV, TXT, Excel (.xlsx, .xls), JSON, Parquet, Feather, HDF5, ORC e Pickle.
    Retorna:
        pd.DataFrame ou None: O DataFrame carregado contendo os dados do arquivo.
    """
    # Se não passar nada por parâmetro, pede o input na tela
    if not caminho_ou_url:
        caminho_ou_url = input("Digite ou cole o caminho/URL do arquivo: ").strip()

    # Remove aspas extras caso o usuário cole o caminho copiado
    caminho_ou_url = caminho_ou_url.strip("'\":")

    # Verifica se é um link web ou arquivo local
    is_url = caminho_ou_url.startswith(('http://', 'https://', 'ftp://'))

    if not is_url and not os.path.exists(caminho_ou_url):
        print(f"❌ Erro: O arquivo local '{caminho_ou_url}' não foi encontrado.")
        return None

    # Extrai a extensão do arquivo (mesmo em URLs)
    extensao = os.path.splitext(caminho_ou_url.split('?')[0])[1].lower()
    df = None

    try:
        if extensao in ['.csv', '.txt']:
            try:
                df = pd.read_csv(caminho_ou_url, **kwargs)
            except Exception:
                df = pd.read_csv(caminho_ou_url, sep=None, engine='python', **kwargs)

        elif extensao == '.tsv':
            df = pd.read_csv(caminho_ou_url, sep='\t', **kwargs)

        elif extensao in ['.xlsx', '.xls']:
            df = pd.read_excel(caminho_ou_url, **kwargs)

        elif extensao == '.json':
            try:
                df = pd.read_json(caminho_ou_url, **kwargs)
            except ValueError:
                df = pd.read_json(caminho_ou_url, lines=True, **kwargs)

        elif extensao in ['.parquet', '.pq']:
            df = pd.read_parquet(caminho_ou_url, **kwargs)

        elif extensao in ['.ftr', '.feather']:
            df = pd.read_feather(caminho_ou_url, **kwargs)

        elif extensao in ['.h5', '.hdf5', '.hdf']:
            df = pd.read_hdf(caminho_ou_url, **kwargs)

        elif extensao == '.orc':
            df = pd.read_orc(caminho_ou_url, **kwargs)

        elif extensao in ['.pkl', '.pickle']:
            df = pd.read_pickle(caminho_ou_url, **kwargs)

        else:
            print(f"⚠️ Formato '{extensao}' não reconhecido. Tentando leitor genérico de texto/CSV...")
            df = pd.read_csv(caminho_ou_url, sep=None, engine='python', **kwargs)

        nome_exibicao = caminho_ou_url.split('/')[-1] if is_url else os.path.basename(caminho_ou_url)
        print(f"✅ Arquivo '{nome_exibicao}' carregado com sucesso! ({df.shape[0]} linhas x {df.shape[1]} colunas)\n")

        # Exibe as 10 primeiras linhas
        display(df.head(10))

        # Retorna o DataFrame para ser guardado numa variável
        return df

    except Exception as e:
        print(f"❌ Erro ao ler o arquivo '{caminho_ou_url}': {e}")
        return None

# Execução ao rodar o script diretamente:
if __name__ == "__main__":
    # A variável 'df' guarda o retorno da função
    df = carregar_arquivos()
```
